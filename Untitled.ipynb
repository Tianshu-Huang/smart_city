```python
import pandas as pd

# ============================================================
# 0. PennDOT Helper — Convert DMS ("40 40:02.398") → Decimal
# ============================================================

def dms_to_decimal(dms_string):
    """
    Convert PennDOT DMS format '40 40:02.398' into decimal degrees.
    """
    try:
        if not isinstance(dms_string, str):
            return None

        parts = dms_string.strip().split(" ")
        deg = float(parts[0])

        min_sec = parts[1].split(":")
        minutes = float(min_sec[0])
        seconds = float(min_sec[1])

        dec = deg + minutes/60 + seconds/3600
        return dec
    except:
        return None

```

```python
# ============================================================
# 1. Load all CSVs
# ============================================================

print("\n=== Loading CSV Files ===")

CRASH = "CRASH_ALLEGHENY_2024.csv"
VEH = "VEHICLE_ALLEGHENY_2024.csv"
RDWY = "ROADWAY_ALLEGHENY_2024.csv"

crash = pd.read_csv(CRASH, low_memory=False)
vehicle = pd.read_csv(VEH, low_memory=False)
roadway = pd.read_csv(RDWY, low_memory=False)

print(f"Crash rows loaded: {len(crash)}")
print(f"Vehicle rows loaded: {len(vehicle)}")
print(f"Roadway rows loaded: {len(roadway)}")
```

```python

# ============================================================
# 2. Apply decimal conversion to lat/lon
# ============================================================

print("\n=== Converting coordinate formats (DMS → decimal) ===")

crash["LAT_DEC"] = crash["LATITUDE"].apply(dms_to_decimal)
crash["LON_DEC"] = crash["LONGITUDE"].apply(dms_to_decimal)

print("Sample converted lat/lon:")
print(crash[["LATITUDE", "LAT_DEC", "LONGITUDE", "LON_DEC"]].head())
```

```python
# ============================================================
# 3. Extract NON-MOTORIST crashes
#   → This includes pedestrians + cyclists (both count as non-motorists)
# ============================================================

print("\n=== Filtering for crashes with pedestrians or cyclists ===")

crash_nonmotr = crash[ crash["NONMOTR_COUNT"] > 0 ].copy()

print(f"Crashes with NONMOTR_COUNT > 0: {len(crash_nonmotr)}")
```

```python

# ============================================================
# 4. Detect INTERSECTION crashes
#   PennDOT definition:
#   INTERSECT_TYPE:
#       00 – Mid-block → NOT intersection
#       01–13 → intersection
#       99 → unknown
# ============================================================

print("\n=== Filtering intersection crashes ===")

crash_nonmotr["IS_INTERSECTION"] = crash_nonmotr["INTERSECT_TYPE"].apply(
    lambda x: int(x != 0 and x != 99)
)

crash_intersections = crash_nonmotr[ crash_nonmotr["IS_INTERSECTION"] == 1 ].copy()

print(f"Intersection-related crashes: {len(crash_intersections)}")
```

```python
# ============================================================
# 5. Detect RIGHT TURN vehicle involvement
#
#   From PennDOT dictionary:
#       VEH_MOVEMENT codes for right turn:
#           09 – Turning right on red
#           10 – Turning right
# ============================================================

print("\n=== Detecting right-turn involved crashes ===")

# Keep only CRN + VEH_MOVEMENT
veh = vehicle[["CRN", "VEH_MOVEMENT"]].copy()

# Right-turn codes
RIGHT_TURN_CODES = {9, 10}

# Determine which CRNs had right-turning units
crn_right_turn = (
    veh[ veh["VEH_MOVEMENT"].isin(RIGHT_TURN_CODES) ]
    .groupby("CRN")
    .size()
    .reset_index(name="RIGHT_TURN_COUNT")
)

print(f"Crashes with right-turning vehicles: {len(crn_right_turn)}")
```

```python

# Merge into crash table
crash_intersections = crash_intersections.merge(
    crn_right_turn[["CRN", "RIGHT_TURN_COUNT"]],
    on="CRN",
    how="left"
)

crash_intersections["RIGHT_TURN_FLAG"] = crash_intersections["RIGHT_TURN_COUNT"].fillna(0).apply(lambda x: 1 if x > 0 else 0)

print("Right turn flag applied.")
```

```python
# ============================================================
# 6. Add roadway name
#   ROADWAY.csv → STREET_NAME
# ============================================================

print("\n=== Adding roadway names ===")

# Keep only CRN + STREET_NAME
rd = roadway[["CRN", "STREET_NAME"]].copy()

# Convert to string, replace NaN with empty string
rd["STREET_NAME"] = rd["STREET_NAME"].fillna("").astype(str)

# Remove blank names so they don't appear in the join
rd = rd[rd["STREET_NAME"].str.strip() != ""]

# Group by CRN
rd = (
    rd.groupby("CRN")["STREET_NAME"]
      .apply(lambda x: "; ".join(sorted(set(x))))
      .reset_index()
)

crash_intersections = crash_intersections.merge(rd, on="CRN", how="left")

print("Roadway names merged successfully.")

```

```python
# ============================================================
# 7. Filter to Strip District bounding box
#
# Strip District APPROX bounding box (you can adjust):
#   Latitude  40.445 – 40.455
#   Longitude -79.996 – -79.967
# ============================================================

print("\n=== Filtering for Strip District crashes ===")

LAT_MIN, LAT_MAX = 40.44, 40.46
LON_MIN, LON_MAX = 79.975, 79.995

crash_strip = crash_intersections[
    (crash_intersections["LAT_DEC"].between(LAT_MIN, LAT_MAX)) &
    (crash_intersections["LON_DEC"].between(LON_MIN, LON_MAX))
].copy()

print(f"Final Strip District crashes: {len(crash_strip)}")
print(crash_intersections)
```
